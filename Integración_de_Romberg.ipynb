```python
import math
import numpy
```

siguiendo los pasos del libro

```python
def Romb(f,a,b,n):
  R1=numpy.zeros(n+1)
  R2=numpy.zeros(n+1)
  Tabla=numpy.zeros((n+1,n+1))
  h=b-a
  R1[1]=0.5*h*(f(a+f(b)))
  Tabla[1,1]=R1[1]
  print(f"R_(1,1)={R1[1]:.9f}")

#aproximacion con regla trapezoidal
  for i in range(2,n+1):
   sum_k=sum(f(a+(k-0.5)*h) for k in range(1,2**(i-2)+1))
   R2[1]=0.5*(R1[1]+h*sum_k)
   Tabla[i,1]=R2[1]
   #extrapolacion
  for j in range(2,i+1):
     R2[j]=R2[j-1]+(R2[j-1]-R1[j-1])/(4**(j-1)-1)
     Tabla[i,j]=R2[j]

     h=h/2.0
     for j in range(1,i+1):
       R1[j]=R2[j]

  return Tabla[1:,1:]
```

Definimos  la funcion del ejemplo 2 de la pagina 160 del libro: $$\int_{0}^{Π} sen(x) \, dx$$ con  a =0, b=Π y n=2

```python
def f(x):
  return math.sin(x)
```

```python
import math
a = 0.0
b = math.pi
n = 6
tabla_res = Romb(f, a, b, n)
print("\n" + "=" * 80)
print("                     TABLA 4.11 DE ROMBERG (BURDEN)")
print("=" * 80)

for i in range(n):
    fila = ""
    for j in range(i + 1):
        fila += f"{tabla_res[i, j]:12.8f} "
    print(fila)

print("=" * 80)
print(f"★ SOLUCIÓN FINAL (R_6,6): {tabla_res[5, 5]:.9f}")
print("=" * 80)
```
